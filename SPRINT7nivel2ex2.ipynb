```python
# DADES ORIGINALES
clientes = [
    {"nom": " Ana ", "edat": 42, "codi": 0},
    {"nom": "Carlos", "edat": 30, "codi": 2},
    {"nom": "Isabel", "edat": 55, "codi": 3},
    {"nom": "Jorge", "edat": 40, "codi": 1},
    {"nom": "Marta", "edat": 28, "codi": 0}
]

niveles_seguro = {
    0: {"categoria": "Baix risc", "preu_base": 120, "fraude": 0.01},
    1: {"categoria": "Risc mitjà", "preu_base": 200, "fraude": 0.05},
    2: {"categoria": "Risc alt", "preu_base": 350, "fraude": 0.15},
    3: {"categoria": "Risc crític", "preu_base": 500, "fraude": 0.30}
}
```

```python
def validar(clientes):

  clientes_v2 = []
  for cliente in clientes:
    cliente_v2 = {}
    cliente_v2['NOMBRE'] = cliente['nom'].strip()
    cliente_v2['EDAT'] = int(cliente['edat'])
    cliente_v2['CODI'] = int(cliente['codi'])
    clientes_v2.append(cliente_v2)
  return clientes_v2

clientes_v2=validar(clientes)
print(clientes_v2)
```

```python
def crear_dict(clientes_v2):
  clientes_v3 = []
  for cliente in clientes_v2:
    cliente_v3 = {}

    cliente_v3['NOMBRE'] = cliente['NOMBRE']
    cliente_v3['EDAT'] = cliente['EDAT']


# LAURA RECUERDA FORMA DE REMPLAZAR CON POR OTRO DICCIONARIO

    datos_seguro = niveles_seguro[cliente['CODI']]

    cliente_v3['CATEGORIA'] = datos_seguro['categoria']
    cliente_v3['PREU_BASE'] = round(datos_seguro['preu_base'])
    cliente_v3['FRAUDE'] = datos_seguro['fraude']

    clientes_v3.append(cliente_v3)
  return clientes_v3

clientes_v3 = crear_dict(clientes_v2)

print(clientes_v3)
```

```python
def calcular_importe(clientes_v3):

  clientes_v4 = []
  for cliente in clientes_v3:
    cliente_v4 = cliente.copy()

    cliente_v4['IMPORTE_FINAL']= round(cliente['PREU_BASE']*(1+cliente['EDAT']/100),2)
    cliente_v4['%_Probabilidad_FRAUDE']=(round(cliente['FRAUDE']*(1-cliente['EDAT']/100),4))*100
    clientes_v4.append(cliente_v4)
  return clientes_v4

clientes_v4 = calcular_importe(clientes_v3)
print(clientes_v4)
```

```python
def crear_dict_clientes(clientes_v4):
  dict_clientes_definitivo = {}
  id_cliente = 0
  for cliente in clientes_v4:
    id_cliente = id_cliente +1
    cliente_v5 = cliente.copy()
    dict_clientes_definitivo[id_cliente ]=cliente_v5

  return dict_clientes_definitivo

dict_clientes_definitivo= crear_dict_clientes(clientes_v4)
print(dict_clientes_definitivo)
```

```python
def ordenar(dict_clientes_definitivo):
  list_clientes_ordenados = sorted(dict_clientes_definitivo.items(), key=lambda item: item [1]['IMPORTE_FINAL'],reverse=True)


  dict_clientes_ordenado = {}
  for cliente in list_clientes_ordenados:
    dict_clientes_ordenado[cliente[0]]=cliente[1]
  return dict_clientes_ordenado

dict_clientes_ordenado = ordenar(dict_clientes_definitivo)
print(dict_clientes_ordenado)

```

```python
def mostrar_dict_clientes(dict_clientes_ordenado):
  print("\n"
  "DICCIONARIO DE CLIENTES ORDENADO POR IMPORTE FINAL DESCENDIENTE\n" )
  resultado= print(dict_clientes_ordenado)
  return resultado
resultado = mostrar_dict_clientes(dict_clientes_ordenado)


```

```python
"""
Un apartat d'anàlisi on s'indiqui:

1 - Quin client té més probabilitat de frau

2 - Quin té el preu final més alt SEGUN RESULTADO CALCULO PREU FINAL

3 - Qui seria prioritari de revisar manualment LOS QUE TIENEN FRAUDE MAS ELEVADO
"""

def respuestas(dict_clientes_ordenado):
    # LAURA MAXIMO DE DICCIONARIO
    cliente_max_fraude = max(
      dict_clientes_ordenado.items(),
      key=lambda item: item[1]['%_Probabilidad_FRAUDE'])

    cliente_max_precio = max(
      dict_clientes_ordenado.items(),
      key=lambda item: item[1]['IMPORTE_FINAL'])

    # cliente_max_fraude es una tupla: (id_cliente, datos_dict)
    # Extraemos el diccionario de datos que está en la segunda posición (índice 1)
    datos_fraude = cliente_max_fraude[1]
    datos_precio = cliente_max_precio[1]

    informe = (
      "\n--- APARTADO DE ANALISIS ---\n"

      f"1 - Cliente con el max_fraude: {datos_fraude['NOMBRE']} Probabilidad: {datos_fraude['%_Probabilidad_FRAUDE']}\n"

      f"2 - Cliente con el precio final más alto: {datos_precio['NOMBRE']} Importe: {datos_precio['IMPORTE_FINAL']}€\n"

      f"3 - Prioritario de revisar manualmente: {datos_fraude['NOMBRE']}\n"
    )
    print(informe)

    return informe

informe = respuestas(dict_clientes_ordenado)

```

```python
def main_estudio_seguros():
  clientes_v2=validar(clientes)
  clientes_v3 = crear_dict(clientes_v2)
  clientes_v4 = calcular_importe(clientes_v3)
  dict_clientes_definitivo= crear_dict_clientes(clientes_v4)
  dict_clientes_ordenado = ordenar(dict_clientes_definitivo)
  resultado = mostrar_dict_clientes(dict_clientes_ordenado)
  informe = respuestas(dict_clientes_ordenado)


```
