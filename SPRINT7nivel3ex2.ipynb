¿Qué hace cada instrucción?

- archivo.keys() obtiene el nombre del archivo que has subido.

- list(...)[0] selecciona el primer archivo.
- open() abre el archivo.

- "r" significa que lo abrimos en modo lectura (read).
- encoding="utf-8" permite leer correctamente tildes y caracteres especiales.

- f.read() lee todo el contenido.

- with permite que el archivo se cierre automáticamente al terminar.

```python
nombre_archivo = list(archivo.keys())[0]

with open(nombre_archivo, "r", encoding="utf-8") as f:
    texto = f.read()
```

```python
def limpiar_lista(texto):
  texto_limpio = texto.strip()

  texto2 = texto_limpio.replace("\n", ",") #quitar saltos

  texto3 = texto2.replace("\t", "_") #quitar tabulaciones


  lista_partidos = texto3.split(",")


  anidacion_lista = []
  for partidos in lista_partidos:
    separar = partidos.replace("_", ",").replace("-", ",").split(",")
    anidacion_lista.append(separar)
  return anidacion_lista

anidacion_lista=limpiar_lista(texto)
```
