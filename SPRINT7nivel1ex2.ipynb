primera parte crear obtenir_segment_i_factor sin condicionalees tipo if o elfi, la opcion primerpo creamos segment con los valores que queremos que tenga, y su clave sera los valores de codi de comportaent, para luego recorre             a usar sera un match case donde creara una nueva clave segment para cada tipo dee comportamente.
ejemplo  "codi_comportament": 4 "segment": alto
ampliar diccionario con nueva clave segment que incluya valores de bajo a alto



```python

def obtenir_segment_i_factor(clients):

  segment = {0: "bajo", 1: "bajo-medio", 2: "medio", 3: "medio-alto", 4: "alto"}
  factor_codi = {0: 0.1, 1: 0.25, 2: 0.5, 3: 0.75, 4: 0.9}
  clients_2=[]
  for cliente in clients:

    # recorremos lista para estar en dict de persona Obtenemos el código de comportamiento del cliente
    codigo = cliente["codi_comportament"]

    # cliente es un diccionaro que ampliamos con Obtenemos el segmento correspondiente desde el diccionario usando el código como clave
    cliente["segment"] = segment[codigo]
    cliente["factor_codi"] = factor_codi[codigo]

    clients_2.append(cliente)
  return clients_2

# Mostramos el resultado para verificar
clients_2= obtenir_segment_i_factor(clients)
print(clients_2)
```

```python
def obtenir_factor_edat(clients_2):

  factor_edad = {"18-25":1.2 ,"26-40":1 ,"41-60":0.85 ,"60+":0.7}
  clients_3=[]

  for cliente in clients_2:
    edades= cliente['grup_edat']
    cliente['factor_edad'] = factor_edad[edades]
    clients_3.append(cliente)


  return clients_3

clients_3 = obtenir_factor_edat(clients_2)
print(clients_3)
```

```python
# he cambiado aqui a diccionario dejando como clave la id_del cliente, si no es correcto y se quiere clave  luego todo seria
# clients_4[cliente['id_client']]= cliente


def calcular_risc(clients_3):

  clients_4 = {}
  for cliente in clients_3:

    risc = round(float(cliente['factor_codi']*cliente['factor_edad']),2)
    cliente["risc_final"]= risc

    clients_4[cliente['id_client']]= {
      "nom":cliente['nom'],
      'edat':cliente['edat'],
      'grup_edat':cliente['grup_edat'],
      'codi_comportament': cliente['codi_comportament'],
      'segment':cliente['segment'],
      'factor_codi':cliente['factor_codi'],
      'factor_edad': cliente['factor_edad'],
      'risc_final' :cliente['risc_final']
      }
  return clients_4

clients_4 = calcular_risc(clients_3)
print(clients_4)
```
