```python
"""
calcular puntuaciones con variables experiencia,catala,excel,titol
aplicant el pes de l'experiència (anys * 1.5) i sumant-hi els punts de català, excel i titulació
"""
def calcular_puntuacio(experiencia,catala,excel,titol):
    puntuacion = experiencia * 1.5 + catala + excel + titol
    return puntuacion

"""
puntuacion = calcular_puntuacio(experiencia,catala,excel,titol)
print("Puntuación:", puntuacion)
"""
```

```python
"""
classificar automàticament cada persona segons els llindars establerts per l'empresa:

» Prioritat Alta → puntuació ≥ 12
» Candidat adequat → 8 ≤ puntuació < 12
» No prioritari → puntuació < 8
"""
def clasificar_candidat(puntuacion):
    if puntuacion >= 12:
        clasificacion = "Prioritat Alta"
    elif puntuacion >= 8:
        clasificacion = "Candidat adequat"
    else:
        clasificacion = "No prioritari"
    return clasificacion


"""
clasificacion = clasificar_candidat(puntuacion)
print(clasificacion)
print("Clasificación:", clasificacion)
"""
```

```python
def mostrar(nombre,telefono,puntuacion,clasificacion):
  print("\n"
      f"el candidato {nombre} tiene una puntuacion de {puntuacion} en calculo de clasificacion y se considera como {clasificacion}\n")
```

```python
def mostrar_informe(lista_candidatos):
  print("\n--- INFORME FINAL DE CANDIDATOS ---")
  print("\n lista de candidatos ordenada por puntuación (de mayor a menor):\n")

  # Ordenamos la lista usando la posición 2 de cada candidato (su puntuación) de mayor a menor (nota Laura ; ESTA PARTE REALIZADA CON IA NO ENCONTRABA COMO HACER SINTAXI PARA ORDENAR POR LA VARIABLE
  lista_ordenada = sorted(lista_candidatos, key=lambda candidato: candidato[2], reverse=True)

  # Mostramos cada candidato ordenadamente
  for candidato in lista_ordenada:
    nombre, telefono, puntuacion, clasificacion = candidato
    print(f"Candidato: {nombre.title()} | Teléfono: {telefono} | Puntuación: {puntuacion} | Clasificación: {clasificacion}")
```

```python
def main_candidatos():
  lista_candidatos = []
  for i in range(3):
    nombre,telefono,experiencia,catala,excel,titol= obtenir_dades()

    nombre,telefono,experiencia,catala,excel,titol= validar_dades(nombre,telefono,experiencia,catala,excel,titol)

    puntuacion = calcular_puntuacio(experiencia,catala,excel,titol)

    clasificacion = clasificar_candidat(puntuacion)

    lista_candidatos.append([nombre,telefono,puntuacion,clasificacion])

    persona = mostrar(nombre,telefono,puntuacion,clasificacion)

  mostrar_informe (lista_candidatos)

  return lista_candidatos
```
