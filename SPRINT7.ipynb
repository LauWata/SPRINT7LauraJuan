```python
"""
calcular puntuaciones con variables experiencia,catala,excel,titol
aplicant el pes de l'experiència (anys * 1.5) i sumant-hi els punts de català, excel i titulació
"""
experiencia=7
catala=3
excel=2
titol=1

def calcular_puntuacio(experiencia,catala,excel,titol):
    resultado = experiencia * 1.5 + catala + excel + titol
    return resultado

puntuacion = calcular_puntuacio(experiencia,catala,excel,titol)
print("Puntuación:", puntuacion)
```

```python
"""
classificar automàticament cada persona segons els llindars establerts per l'empresa:

» Prioritat Alta → puntuació ≥ 12
» Candidat adequat → 8 ≤ puntuació < 12
» No prioritari → puntuació < 8
"""

puntuacion=11

def clasificar_candidat(puntuacion):
    if puntuacion >= 12:
        clasificacion = "Prioritat Alta"
    elif puntuacion >= 8:
        clasificacion = "Candidat adequat"
    else:
        clasificacion = "No prioritari"
    return clasificacion

clasificacion = clasificar_candidat(puntuacion)
print(clasificacion)
print("Clasificación:", clasificacion)
```
