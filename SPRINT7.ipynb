```python
# funcion formulaio para que usuario inserte los datos de los candidatos entiendo que al ser formulario se debe usar input para insertar de manera manual

def obtenir_dades():
  nombre = input("nombre y apellido")
  telf = input(
    "inserta num de telefon,: \n"
    "telefono formato sin espacios ej;666221155 "
    )
  exp= int(input("Anys d'experiència dins de l'empresa"))
  catala = int(input(
    "Indica el nivel de catalán:\n"
    "0 = Sin conocimientos\n"
    "1 = Hasta nivel B1\n"
    "2 = Nivel B2\n"
    "3 = C1 o superior\n"
    "Tu respuesta: "))
  excel = int(input(
    "Indica el nivel de excel:\n"
    "0 = nivell bàsic\n"
    "1 = ús de funcions B1\n"
    "2 = taules dinàmiques B2\n"
    "3 = Visual Basic i Macros\n"
    "Tu respuesta: "))
  titol = int(input(
    "Indica tu nivel academic: \n"
    "0 = ESO \n"
    "1 = FP \n"
    "2 = Universitat \n"
    "3 = Màster o superior\n"
  ))

  datos_personales = [nombre, telf, exp, catala,excel,titol]
  return datos_personales

print("hola asistente para insertar datos:\n"
      "inserta datos de primer candidato"
      )
persona1 = obtenir_dades()
print("ahora sigamos con datos candidato 2")
persona2 = obtenir_dades()
print("ahora sigamos con datos candidato 3")
persona3 = obtenir_dades()
print("ahora sigamos con datos candidato 4")
persona4 = obtenir_dades()


print(persona1)
print(persona2)
print(persona3)
print(persona4)
```

```python
# Restauramos la función print original en el entorno
import builtins
print = builtins.print

notas = [4, 3, 5, 8, 2, 8, 9]
aprovado = []
suspensos = []

for nota in notas:
    if nota >= 5:
        aprovado.append(nota)
    else:
        suspensos.append(nota)

print("hay un total de,", len(aprovado), "aprovados")
print("hay un total de,", len(suspensos), "suspensos")
```
