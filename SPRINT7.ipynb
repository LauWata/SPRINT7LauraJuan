LA PRIMERA FUNCION SERA UNA QUE FABRIQUE UN FORMULARIO, NO RECIBE NADA Y DEVOLVERA UNA LISTA DE DATOS DE UNMA PERSONA

```python
def obtenir_dades():
  print("hola asistente para insertar datos:\n"
      "inserta datos de primer candidato"
      )
  nombre = input("nombre y apellido").strip()

  telf = input(
    "inserta num de telefon,: \n"
    "telefono formato sin espacios ej;666221155 "
    )

  exp= int(input("Anys d'experiència dins de l'empresa"))


  catala = int(input(
    "Indica el nivel de catalán:\n"
    "0 = Sin conocimientos\n"
    "1 = Hasta nivel B1\n"
    "2 = Nivel B2\n"
    "3 = C1 o superior\n"
    "Tu respuesta: "))
  excel = int(input(
    "Indica el nivel de excel:\n"
    "0 = nivell bàsic\n"
    "1 = ús de funcions B1\n"
    "2 = taules dinàmiques B2\n"
    "3 = Visual Basic i Macros\n"
    "Tu respuesta: "))
  titol = int(input(
    "Indica tu nivel academic: \n"
    "0 = ESO \n"
    "1 = FP \n"
    "2 = Universitat \n"
    "3 = Màster o superior\n"
  ))
  datos_personales = [nombre, telf, exp, catala,excel,titol]
  return datos_personales

```
