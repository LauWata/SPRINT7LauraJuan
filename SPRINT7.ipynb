```python
#VALIDACION DE NOMBRE QUE NO ESTE VACIO Y QUE NO TENGA NUMERO NI CARACTERES

nombre= input("dime nombre")
def validar_nombre_con_try(nombre):
  while True:
    if not nombre.replace(" ", "").isalpha() or nombre.strip() == "":
      print("El nombre contiene caracteres no permitidos o está vacío")
      nombre=input("vuelva a escribir su nombre")
    else:
      print("nombre registtrado")
      return nombre

# Probamos la función
nombre_correcto = validar_nombre_con_try(nombre)
print( nombre_correcto)
```
