```python
while True:
  persona = input("dime nombre: ")
  # Eliminamos espacios temporalmente para validar que solo haya letras
  if persona.replace(" ", "").isalpha() and persona.strip()
  != "":
    break
  else:
    print("Valor no válido. Por favor, introduce un nombre que contenga solo letras.")

print("nombre registrado:", persona)
```
