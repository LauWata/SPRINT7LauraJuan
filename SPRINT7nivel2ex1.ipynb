```python
pacients = [

"Maria|45|tos, Fiebre Alta, dolor_pecho",

"Luis|33|dolor_cabeza,faTiga",

"Sara|67|fiebre alta, dificultad RESPIRAR, tos",

"Jordi|52|fatiga , tos",

"Anna|29|fiebre alta,dolor_cabeza" ]


nivells_risc = {

"fiebre_alta": 3,

"dificultad_respirar": 5,

"dolor_pecho": 4,

"tos": 1,

"fatiga": 1,

"dolor_cabeza": 1 }
```

```python
#funcion limpieza de datos de la lista
def limpiar(pacients):
  lista_pacientes=[]
  for texto in pacients:
    texto_limpio=texto.lower().replace("_"," ").strip()
    lista_pacientes.append(texto_limpio)
  return lista_pacientes
lista_pacientes=limpiar(pacients)
print(lista_pacientes)
```

```python
#funcion separar la lista por pacientes
def separar(lista_pacientes):
  lista_pacientesv2=[]
  for texto in lista_pacientes:

    texto2= texto.split("|") # separar mediante split no sabia que split

    lista_pacientesv2.append(texto2)
  return lista_pacientesv2

lista_pacientesv2=separar(lista_pacientes)
print(lista_pacientesv2)
```

```python
#funcion crear lista pacientes y un diccionario por paciente, con las claves nombre, edad, y el resumen de los sintomas
def crear_dic_pacientte(lista_pacientesv2):

  lista_pacientesv3 = []
  for pacinete in lista_pacientesv2:
    dict_paciente={}
    dict_paciente["nombre"]=pacinete[0]
    dict_paciente["edad"]=pacinete[1]
    dict_paciente["resum_sintomas"]=pacinete[-1]
    lista_pacientesv3.append(dict_paciente)

  return lista_pacientesv3
lista_pacientesv3= crear_dic_pacientte(lista_pacientesv2)
print(lista_pacientesv3)
```
