```python
import random
```

```python
def definir_longitud():
    longitud_variable = random.randint(10, 20)
    return longitud_variable

#mostrar funcion
longitud_variable = definir_longitud()
print(longitud_variable)
```

```python
def definir_caracteres(longitud_variable, mayusculas=True, minusculas=True, numeros=True, signos=True):
  grupo_mayusculas="ABCDEFGHIJKLMNÑOPQRSTUVWXYZ"
  grupo_minusculas = "abcdefghijklmnñopqrstuvwxyz"
  grupo_numeros = "0123456789"
  grupo_signos = "!@#$%&*?"

  lista_permitidos = []

  if mayusculas:
    lista_permitidos.extend(grupo_mayusculas)

  if minusculas:
    lista_permitidos.extend(grupo_minusculas)

  if numeros:
    lista_permitidos.extend(grupo_numeros)

  if signos:
    lista_permitidos.extend(grupo_signos)

  contrasenia = []

  for i in range(longitud_variable):
    nuevos = random.choice(lista_permitidos)
    contrasenia.append(nuevos)

  random.shuffle(contrasenia)
  return contrasenia

#mostrar funcion
contrasenia = definir_caracteres(longitud_variable,mayusculas=True, minusculas=True, numeros=True, signos=True)
print(contrasenia)
```

```python
def mostrar_contrasenia(contrasenia):
  resultado="".join(contrasenia)
  return resultado
resultado=mostrar_contrasenia(contrasenia)
print(resultado)
```

```python
def main_crear_contrasenia():
  longitud_variable = definir_longitud()
  contrasenia = definir_caracteres(longitud_variable,mayusculas=True, minusculas=True, numeros=True, signos=True)
  resultado=mostrar_contrasenia(contrasenia)
  print(resultado)

```

```python
main_crear_contrasenia()
```
