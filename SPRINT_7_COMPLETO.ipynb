primera parte crear obtenir_segment_i_factor sin condicionalees tipo if o elfi, la opcion primerpo creamos segment con los valores que queremos que tenga, y su clave sera los valores de codi de comportaent, para luego recorre             a usar sera un match case donde creara una nueva clave segment para cada tipo dee comportamente.
ejemplo  "codi_comportament": 4 "segment": alto
ampliar diccionario con nueva clave segment que incluya valores de bajo a alto



```python

def obtenir_segment_i_factor(clients):

  segment = {0: "bajo", 1: "bajo-medio", 2: "medio", 3: "medio-alto", 4: "alto"}
  factor_codi = {0: 0.1, 1: 0.25, 2: 0.5, 3: 0.75, 4: 0.9}
  clients_2=[]
  for cliente in clients:

    # recorremos lista para estar en dict de persona Obtenemos el código de comportamiento del cliente
    codigo = cliente["codi_comportament"]

    # cliente es un diccionaro que ampliamos con Obtenemos el segmento correspondiente desde el diccionario usando el código como clave
    cliente["segment"] = segment[codigo]
    cliente["factor_codi"] = factor_codi[codigo]

    clients_2.append(cliente)
  return clients_2

# Mostramos el resultado para verificar
clients_2= obtenir_segment_i_factor(clients)
print(clients_2)
```

```python
#funcion limpieza de datos de la lista
def limpiar(pacients):
  lista_pacientes=[]
  for texto in pacients:
    texto_limpio=texto.lower().replace("_"," ").replace(",","|").strip()
    lista_pacientes.append(texto_limpio)
  return lista_pacientes
lista_pacientes=limpiar(pacients)
print(lista_pacientes)
```

```python
#funcion separar la lista por pacientes
def separar(lista_pacientes):
  lista_pacientesv2=[]
  for paciente in lista_pacientes:
    paciente=paciente.strip()
    pacientev2= paciente.split("|") # separar mediante split no sabia que split
    pacientev3 = []

    for valores in pacientev2:
      valor_limpio = valores.strip().replace(" ","_")
      pacientev3.append(valor_limpio)
    lista_pacientesv2.append(pacientev3)
  return lista_pacientesv2

lista_pacientesv2=separar(lista_pacientes)

print(lista_pacientesv2)

```

```python
#crear un diccionario con id pacientes y sus sintomas puntuados

#NOTA DE LAURA SIME DA TIEMPO MIRARE DE MEJORAR ESTO DE TANTO IF

def crear_dic_pacientes(lista_pacientesv2):
  dic_todospacientes = {}
  id_paciente = 0

  for paciente in lista_pacientesv2:
    dic_paciente ={}

    dic_paciente["Nombre"]= paciente[0]

    dic_paciente["Edad"]= paciente[1]

    dic_paciente["Resum:sintomas"]= paciente[2:]

    if "fiebre_alta" in paciente:
      dic_paciente["fiebre_alta"] = nivells_risc['fiebre_alta']
    else:
      dic_paciente["fiebre_alta"] = 0
    if "dolor_pecho" in paciente:
      dic_paciente["dolor_pecho"] = nivells_risc['dolor_pecho']
    else:
      dic_paciente["dolor_pecho"] = 0
    if "dolor_cabeza" in paciente:
      dic_paciente["dolor_cabeza"] = nivells_risc['dolor_cabeza']
    else:
      dic_paciente["dolor_cabeza"] = 0
    if "dificultad_respirar" in paciente:
      dic_paciente["dificultad_respirar"] = nivells_risc['dificultad_respirar']
    else:
      dic_paciente["dificultad_respirar"] = 0
    if "tos" in paciente:
      dic_paciente["tos"] = nivells_risc['tos']
    else:
      dic_paciente["tos"] = 0
    if "fatiga" in paciente:
      dic_paciente["fatiga"] = nivells_risc['fatiga']
    else:
      dic_paciente["fatiga"] = 0

    id_paciente = id_paciente + 1
    dic_todospacientes[id_paciente]=dic_paciente

  return dic_todospacientes

dic_todospacientes = crear_dic_pacientes(lista_pacientesv2)
print(dic_todospacientes)
```

```python
# CALCULAR RIESGO Y CREAR EL DICCIONARIO FINAL CON LA VISION DESEADA


def calcular_risc(dic_todospacientes):

  dic_pacientes_final = {}
  for id_paciente,paciente in dic_todospacientes.items():
    paciente_final = {}

    calcul_risc= paciente['dificultad_respirar']+paciente['dolor_cabeza']+paciente['dolor_pecho']+paciente['fatiga']+paciente['fiebre_alta']+paciente['tos']

    paciente_final["NOMBRE"]=paciente['Nombre']
    paciente_final["EDAD"]=paciente['Edad']
    paciente_final["SINTOMATOLOGIA"]=paciente['Resum:sintomas']
    paciente_final["RISC_TOTAL"]=calcul_risc

    dic_pacientes_final[id_paciente]=paciente_final
  return dic_pacientes_final

dic_pacientes_final=calcular_risc(dic_todospacientes)

print(dic_pacientes_final)

```

```python
# funcion ordenar el diccionario

def ordenar(dic_pacientes_final):

  lista_pacientes_final_ordenado = sorted(dic_pacientes_final.items(), key=lambda item: item[1]['RISC_TOTAL'], reverse=True)


  dic_pacientes_final_ordenado = {}
  for paciente in lista_pacientes_final_ordenado:

    dic_pacientes_final_ordenado[paciente[0]] = paciente [1]


  return dic_pacientes_final_ordenado

dic_pacientes_final_ordenado = ordenar(dic_pacientes_final)
print(dic_pacientes_final_ordenado)

```

```python
def mostrar_informe_prioritarios(dic_pacientes_final_ordenado):
  print("\n"
        "-----------------------------------------------\n"
        "INFORME LISTA PACIENTES POR ORDEN DE PRIORIDAD \n"
         "-----------------------------------------------\n")
  for id,paciente in dic_pacientes_final_ordenado.items():
    informe = print("\n"
        "DATOS DEL PACIENTE\n"
        "-----------------------------------------------\n"
        f"NOMBRE; {paciente['NOMBRE']}\n"
      f"EDAD; {paciente['EDAD']}\n"
      f"SINTOMATOLOGIA;  {paciente['SINTOMATOLOGIA']}\n"
      f"RISC_TOTAL; {paciente['RISC_TOTAL']}")

  return informe

informe = mostrar_informe_prioritarios(dic_pacientes_final_ordenado)
print(informe)


```

```python
def validar(clientes):

  clientes_v2 = []
  for cliente in clientes:
    cliente_v2 = {}
    cliente_v2['NOMBRE'] = cliente['nom'].strip()
    cliente_v2['EDAT'] = int(cliente['edat'])
    cliente_v2['CODI'] = int(cliente['codi'])
    clientes_v2.append(cliente_v2)
  return clientes_v2

clientes_v2=validar(clientes)
print(clientes_v2)
```

```python
def crear_dict(clientes_v2):
  clientes_v3 = []
  for cliente in clientes_v2:
    cliente_v3 = {}

    cliente_v3['NOMBRE'] = cliente['NOMBRE']
    cliente_v3['EDAT'] = cliente['EDAT']


# LAURA RECUERDA FORMA DE REMPLAZAR CON POR OTRO DICCIONARIO

    datos_seguro = niveles_seguro[cliente['CODI']]

    cliente_v3['CATEGORIA'] = datos_seguro['categoria']
    cliente_v3['PREU_BASE'] = round(datos_seguro['preu_base'])
    cliente_v3['FRAUDE'] = datos_seguro['fraude']

    clientes_v3.append(cliente_v3)
  return clientes_v3

clientes_v3 = crear_dict(clientes_v2)

print(clientes_v3)
```

```python
def calcular_importe(clientes_v3):

  clientes_v4 = []
  for cliente in clientes_v3:
    cliente_v4 = cliente.copy()

    cliente_v4['IMPORTE_FINAL']= round(cliente['PREU_BASE']*(1+cliente['EDAT']/100),2)
    cliente_v4['%_Probabilidad_FRAUDE']=(round(cliente['FRAUDE']*(1-cliente['EDAT']/100),4))*100
    clientes_v4.append(cliente_v4)
  return clientes_v4

clientes_v4 = calcular_importe(clientes_v3)
print(clientes_v4)
```

```python
def crear_dict_clientes(clientes_v4):
  dict_clientes_definitivo = {}
  id_cliente = 0
  for cliente in clientes_v4:
    id_cliente = id_cliente +1
    cliente_v5 = cliente.copy()
    dict_clientes_definitivo[id_cliente ]=cliente_v5

  return dict_clientes_definitivo

dict_clientes_definitivo= crear_dict_clientes(clientes_v4)
print(dict_clientes_definitivo)
```

```python
def ordenar(dict_clientes_definitivo):
  list_clientes_ordenados = sorted(dict_clientes_definitivo.items(), key=lambda item: item [1]['IMPORTE_FINAL'],reverse=True)


  dict_clientes_ordenado = {}
  for cliente in list_clientes_ordenados:
    dict_clientes_ordenado[cliente[0]]=cliente[1]
  return dict_clientes_ordenado

dict_clientes_ordenado = ordenar(dict_clientes_definitivo)
print(dict_clientes_ordenado)

```

```python
# esta funcion define los caracteres que tendra la contraseña


def definir_caracteres(longitud_variable, mayusculas=False, minusculas=True, numeros=True, signos=True):
  grupo_mayusculas="ABCDEFGHIJKLMNÑOPQRSTUVWXYZ"
  grupo_minusculas = "abcdefghijklmnñopqrstuvwxyz"
  grupo_numeros = "0123456789"
  grupo_signos = "!@#$%&*?"

  obligatorios = []
  if mayusculas:
    obligatorios.append(random.choice(list(grupo_mayusculas)))
  if minusculas:
    obligatorios.append(random.choice(list(grupo_minusculas)))
  if numeros:
    obligatorios.append(random.choice(list(grupo_numeros)))
  if signos:
    obligatorios.append(random.choice(list(grupo_signos)))



  lista_permitidos = list(grupo_mayusculas)+list(grupo_minusculas)+list(grupo_numeros)

  contrasenia = obligatorios.copy()

  for i in range(longitud_variable-len(obligatorios)):

    nuevos = random.choice(lista_permitidos)
    contrasenia.append(nuevos)

  random.shuffle(contrasenia)
  return contrasenia


#mostrar funcion
contrasenia = definir_caracteres(longitud_variable,mayusculas=False, minusculas=True, numeros=True, signos=True)
print(contrasenia)
```

```python
def mostrar_contrasenia(contrasenia):
  resultado="".join(contrasenia)
  print(resultado)
  return resultado
resultado=mostrar_contrasenia(contrasenia)

```

```python
def main_crear_contrasenia():
  longitud_variable = definir_longitud()
  contrasenia = definir_caracteres(longitud_variable,mayusculas=False, minusculas=True, numeros=True, signos=True)
  resultado=mostrar_contrasenia(contrasenia)


```

```python
def limpiar_lista(texto):
  texto_limpio = texto.strip()

  texto2 = texto_limpio.replace("\n", ",") #quitar saltos

  texto3 = texto2.replace("\t", "_") #quitar tabulaciones


  lista_partidos = texto3.split(",")


  anidacion_lista = []
  for partidos in lista_partidos:
    separar = partidos.replace("_", ",").replace("-", ",").split(",")
    anidacion_lista.append(separar)
  return anidacion_lista

anidacion_lista=limpiar_lista(texto)
```

```python
#YUNIER AQUI TODO CON AYUDA DE IA,
# EN ESTE APARTADO VAMOS A MONTAR EL DICCIONARIO CON LOS RESULTADOS DE CADA EQUIPO, LA SINTAXI PARA CREAR RELLENAR LOS RESULTADOS CONSULTADO CON IA LOS APARTADOS 2 Y 3 PUESTO NO SOY CAPAZ DE MONTAR Y RELLENAR EL DICCIONARIO, AUN NO ENTIENDO LA LOGICA DEL 2


def crear_dicclasificacion(anidacion_lista):
  clasificacion = {}

  for partido in anidacion_lista:

    equipo_local = partido[0]
    equipo_visitante = partido[-1]
    goles_local = int(partido[1])
    goles_visitante = int(partido[2])

    if equipo_local not in clasificacion:
       clasificacion[equipo_local] = {
           "gols_favor": 0,
           "gols_contra": 0,
           "guanyats": 0,
           "empatats": 0,
           "perduts": 0,
           "punts": 0}

    if equipo_visitante not in clasificacion:
       clasificacion[equipo_visitante] = {
           "gols_favor": 0,
           "gols_contra": 0,
           "guanyats": 0,
           "empatats": 0,
           "perduts": 0,
           "punts": 0}

    # 2. ACTUALIZAR LOS GOLES ESTA PARTE GENERADA CON IA, NO SE COMO INCLUIR LOS GOLES DE CADA EQUIPO,
    clasificacion[equipo_local]["gols_favor"] += goles_local
    clasificacion[equipo_local]["gols_contra"] += goles_visitante

    clasificacion[equipo_visitante]["gols_favor"] += goles_visitante
    clasificacion[equipo_visitante]["gols_contra"] += goles_local

     # 3. ACTUALIZAR VICTORIAS, EMPATES Y DERROTAS

    if goles_local > goles_visitante:

      clasificacion[equipo_local]["guanyats"] += 1
      clasificacion[equipo_visitante]["perduts"] += 1
      clasificacion[equipo_local]["punts"] += 3

    elif goles_local < goles_visitante:

      clasificacion[equipo_visitante]["guanyats"] += 1
      clasificacion[equipo_local]["perduts"] += 1
      clasificacion[equipo_visitante]["punts"] += 3

    else:

      clasificacion[equipo_local]["empatats"] += 1
      clasificacion[equipo_visitante]["empatats"] += 1
      clasificacion[equipo_local]["punts"] += 1
      clasificacion[equipo_visitante]["punts"] += 1
  return clasificacion
clasificacion = crear_dicclasificacion(anidacion_lista)
print(clasificacion)
```
